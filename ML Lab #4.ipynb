# Lab 04 - Simple Linear Regression (from scratch)

Theme: Dragon Training Academy 🐉

X = Weekly training hours

Y = Fire-breath distance (meters)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("dragon_training.csv")
X = dataset["TrainingHoursPerWeek"].values
Y = dataset["FireBreathDistance_m"].values
n = len(X)
```

```python
sum_x, sum_y = np.sum(X), np.sum(Y)
sum_xy = np.sum(X * Y)
sum_x2 = np.sum(X ** 2)
```

```python
b1 = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x ** 2)
b0 = (sum_y - b1 * sum_x) / n
```

```python
print(f"Slope (b1):     {b1:.4f}  -> extra meters of fire-breath per extra hour/week")
print(f"Intercept (b0): {b0:.4f}  -> baseline distance at 0 training hours")
```

```python
Y_pred = b0 + b1 * X
```

```python
residuals = Y - Y_pred
mse = np.mean(residuals ** 2)
rmse = np.sqrt(mse)
ss_res = np.sum(residuals ** 2)
ss_tot = np.sum((Y - np.mean(Y)) ** 2)
r2 = 1 - ss_res / ss_tot
```

```python
print(f"MSE:  {mse:.3f}")
print(f"RMSE: {rmse:.3f}")
print(f"R^2:  {r2:.4f}")
```

```python
new_hours = np.array([3, 6.5, 12])
new_preds = b0 + b1 * new_hours
for h, p in zip(new_hours, new_preds):
    print(f"A dragon training {h} hrs/week -> predicted fire-breath distance: {p:.2f} m")
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X, Y, color="crimson", label="Actual dragons", s=70, edgecolor="black", zorder=3)
x_line = np.linspace(X.min(), X.max(), 100)
plt.plot(x_line, b0 + b1 * x_line, color="navy", linewidth=2, label="Fitted regression line")
plt.scatter(new_hours, new_preds, color="gold", marker="*", s=250,
            edgecolor="black", label="New predictions", zorder=4)
plt.title("Dragon Training Academy: Hours Trained vs Fire-Breath Distance", fontsize=13)
plt.xlabel("Training Hours per Week")
plt.ylabel("Fire-Breath Distance (m)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("dragon_regression_plot.png", dpi=150)
print("Plot saved.")
plt.show()
```
